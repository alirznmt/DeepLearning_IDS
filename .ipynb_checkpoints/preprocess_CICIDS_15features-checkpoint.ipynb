```python
import ipaddress
from sklearn.model_selection import train_test_split
import pandas as pd
from sklearn.preprocessing import StandardScaler, LabelEncoder
import pickle
```

```python
df = pd.read_csv('/home/fnst/Downloads/final_data.csv')
df
```

This section performs data preprocessing, including converting IP addresses to numerical values, encoding attack labels into binary format, and normalizing selected numerical features to ensure consistency for model training

```python
df['Src IP'] = df['Src IP'].apply(lambda x: int(ipaddress.IPv4Address(x)))
df['Dst IP'] = df['Dst IP'].apply(lambda x: int(ipaddress.IPv4Address(x)))

# Step 2: Encode the 'Label' column (attack type) to numeric values
label_encoder = LabelEncoder()
df['Label'] = label_encoder.fit_transform(df['Label'])
df['Label'] = df['Label'].apply(lambda x: 1 if x != 0 else 0)

# Step 3: Normalize numerical columns
scaler = StandardScaler()
features = ['Src IP', 'Dst IP', 'Src Port', 'Dst Port', 'Protocol',
                'Total Fwd Packet', 'Packet Length Min', 'Packet Length Max',
                'Packet Length Mean', 'FIN Flag Count', 'SYN Flag Count',
                'RST Flag Count', 'PSH Flag Count', 'ACK Flag Count', 'URG Flag Count']

df[features] = scaler.fit_transform(df[features])
df
```

```python
train_data, test_data = train_test_split(df, test_size=0.2, random_state=42)
train_data
```

This section separates the label column from the feature set in the training data. The processed features and labels are then saved as pickle files for efficient storage and later use in model training.

```python
train_y = train_data['Label']
train_x = train_data.drop(columns=['Label'])
# Save the filtered training data to a pickle file
with open('train_cnn_x.pkl', 'wb') as f:
    pickle.dump(train_x.values, f)
# Save the filtered training data to a pickle file
with open('train_cnn_y.pkl', 'wb') as f:
    pickle.dump(train_y.values, f)

```

This section filters the training data to retain only normal (Label = 0) instances. The extracted features are then saved as a pickle file, which can be used for training an unsupervised anomaly detection model.

```python
train_data_filtered = train_data[train_data['Label'] == 0]
train_x1 = train_data_filtered.drop(columns=['Label'])
# Save the filtered training data to a pickle file
with open('diff_train.pkl', 'wb') as f:
    pickle.dump(train_x1.values, f)
```

This section separates the labels from the feature set in the test data. The processed test features and labels are then saved as pickle files for later use in model evaluation.

```python
test_y = test_data['Label']
test_x = test_data.drop(columns=['Label'])
# Save the filtered training data to a pickle file
with open('test_x.pkl', 'wb') as f:
    pickle.dump(test_x.values, f)
# Save the filtered training data to a pickle file
with open('test_y.pkl', 'wb') as f:
    pickle.dump(test_y.values, f)
```
